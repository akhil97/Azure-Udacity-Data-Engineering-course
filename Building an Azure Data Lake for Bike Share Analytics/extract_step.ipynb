```python
# Read CSV files from the filestore volume and save as Delta tables in bronze schema
volume_path = "/Volumes/databricks_301775/default/filestore/"
catalog_schema = "databricks_301775.bronze"


df = spark.read.format("csv")\
    .option("inferSchema", "false")\
    .option("header", "false")\
    .option("sep", ",")\
    .csv(f"{volume_path}payments.csv") \
    .toDF("payment_id", "date", "amount", "rider_id")

    
df.write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{catalog_schema}.payments")

print(f"✓ payments: {df.count()} rows saved to {catalog_schema}.payments")
```
