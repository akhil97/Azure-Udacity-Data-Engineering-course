```python
# Create SQL tables in silver schema by reading from bronze Delta store using Spark SQL

tables = ["payments", "riders", "stations", "trips"]

for table in tables:
    spark.sql(f"""
        CREATE OR REPLACE TABLE databricks_301775.silver.{table}
        AS SELECT * FROM databricks_301775.bronze.{table}
    """)
    
    count = spark.sql(f"SELECT COUNT(*) as cnt FROM databricks_301775.silver.{table}").collect()[0]["cnt"]
    print(f"✓ databricks_301775.silver.{table} created with {count} rows")
```
